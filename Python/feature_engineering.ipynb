```python
import pandas as pd 
import numpy as np
pd.set_option('display.max_columns', None)
df = pd.read_csv(r"C:\Users\PUNEET\OneDrive\Desktop\deeedidy\github\HR_newwwwwwwwwwwwwwwwwwwwwww\HR_Analytics_Cleaned_Master.csv")
```

```python
df['AgeGroup'].unique()
```

```python
df['AgeGroup'] = pd.cut(
    df['Age'],
    bins=[17,25,35,45,55,65],
    labels=['18-25','26-35','36-45','46-55','55+']
)
```

```python
df['AgeGroup'].value_counts()
```

```python
df['AgeGroup'].value_counts().sort_index()
```

```python
df['MonthlyIncome'].value_counts()
```

```python
df['SalaryBand'] = pd.cut(
    df['MonthlyIncome'],
    bins=[0,30000,60000,100000,200000],
    labels=['Low','Medium','High','Very High']
)
```

```python
df['SalaryBand'].value_counts()
```

```python
df['YearsAtCompany'].value_counts()
```

```python
df['TenureGroup'] = pd.cut(
    df['YearsAtCompany'],
    bins=[-1,2,5,10,20,40],
    labels=[
        'New Hire',
        'Junior',
        'Mid-Level',
        'Senior',
        'Veteran'
    ]
)
```

```python
df['TenureGroup'].value_counts()
```

```python
df['TotalWorkingYears'].value_counts()
```

```python
df['ExperienceLevel'] = pd.cut(
    df['TotalWorkingYears'],
    bins=[-1,5,10,20,40],
    labels=[
        'Entry',
        'Mid',
        'Experienced',
        'Expert'
    ]
)
```

```python
df['ExperienceLevel'].value_counts()
```

```python
df['DistanceCategory'] = pd.cut(
    df['DistanceFromHome'],
    bins=[0,5,10,20,30],
    labels=[
        'Very Near',
        'Near',
        'Moderate',
        'Far'
    ]
)
```

```python
df['DistanceCategory'].value_counts()
```

```python
df['OverallSatisfaction'] = (
    df['EnvironmentSatisfaction'] +
    df['JobSatisfaction'] +
    df['WorkLifeBalance']
)/3
```

```python
df['PromotionDue'] = np.where(
    (df['YearsAtCompany'] >= 5) &
    (df['YearsSinceLastPromotion'] >= 5),
    'Yes',
    'No'
)
```

```python
df['FrequentJobChanger'] = np.where(
    df['NumCompaniesWorked'] >= 5,
    'Yes',
    'No'
)
```

```python
df['IncomePerJobLevel'] = (
    df['MonthlyIncome'] /
    df['JobLevel']
).round(2)
```

```python
df.to_csv(
    "HR_Analytics_Cleaned.csv",
    index=False
)
```
